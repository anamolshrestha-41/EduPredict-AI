## Data Loading

```python
import pandas as pd
import plotly.express as px

df= pd.read_csv("../data/student_dataset/student-mat.csv", sep=";")
print(f"{df.head(5)}, {df.shape}, {df.info()}")
```

## Inspect Missing And Duplicated Values

```python
df.isna().sum()
df.duplicated().sum()
```

## Numerical and Categorical Columns

```python
numerical_cols= df.select_dtypes(include="number").columns
print(f"Numerical Columns: {numerical_cols.tolist()}")
categorical_cols=df.select_dtypes(include=["object", "category"]).columns
print(f"Categorical Columns: {categorical_cols.tolist()}")
```

## Summary Statistics

```python
df.describe()
df.describe(include="object")
df.describe(include="all")
```

## Explore the sidtribution of final grades

```python
fig = px.histogram(
    df,
    x="G3",
    title="Distribution of Final Grades",
    labels={
        "G3": "Final Grade",
        "count": "Number of Students"
    }
)
fig.show()
```

## Study time against final grade

```python
fig = px.box(
    df,
    x="studytime",
    title="Study time vs Final Grade",
    labels={
        "studytime":"Weekly study time category",
        "G3": "Final Grade",
    }
)
fig.show()
```

## Absence and final grades

```python
fig = px.histogram(
    df,
    x="absences",
    title="Distribution of Student Absence",
    labels={
        "absences": "Number of Absence",
        "count": "Number of Students"
    }
)
fig.show()
fig = px.scatter(
    df,
    x="absences",
    title="Student Absences Vs Final Grades",
    labels={
        "absences": "Number of Absences",
        "G3": "Final Grade",
    }
)
fig.show()
df[["absences", "G3"]].corr()
```

## Previous Failures and grades

```python
fig = px.box(
    df,
    x="failures",
    title="Previous Failures vs Final Grades",
    labels={
        "failures": "Number of Previous Failures",
        "G3": "Final Grade"
        }
)
fig.show()
```

## Relationship between G1, G2, G3

```python
grade_corr=df[["G1", "G2", "G3"]].corr()
fig=px.imshow(
    grade_corr,
    title="Correlation between period grades"
)
fig.show()
```
